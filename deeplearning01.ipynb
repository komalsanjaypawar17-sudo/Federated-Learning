```python
import numpy as np
from keras.datasets import mnist
from keras.layers import Dense, Activation, Flatten, Conv2D, MaxPooling2D
from keras.models import Sequential
from keras.utils import to_categorical

# Load data
(train_x, train_y), (test_x, test_y) = mnist.load_data()

# Prepare images
train_x = train_x.reshape(-1, 28, 28, 1).astype('float32') / 255.0
test_x = test_x.reshape(-1, 28, 28, 1).astype('float32') / 255.0
train_y_one_hot = to_categorical(train_y)
test_y_one_hot = to_categorical(test_y)

# Build the model
model = Sequential()
model.add(Conv2D(64, (3, 3), input_shape=(28, 28, 1)))
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Flatten())
model.add(Dense(64, activation='relu'))
model.add(Dense(10))
model.add(Activation('softmax'))

# Train and test
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(train_x, train_y_one_hot, batch_size=128, epochs=10)
test_loss, test_acc = model.evaluate(test_x, test_y_one_hot)
print('Test loss:', test_loss)
print('Test accuracy:', test_acc)

predicted_labels = np.argmax(model.predict(test_x), axis=1)
```
